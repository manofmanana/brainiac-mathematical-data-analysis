# Backpropagation for a Dense Layer — Derivation and Demo

This notebook derives the backward pass for a Dense layer and runs a tiny demo that exports figures into `notebooks/figures/`.

```python
# Setup imports
import sys
from pathlib import Path
PROJ = Path('..').resolve()
sys.path.insert(0, str(PROJ))
import numpy as np
from core.network import NeuralNetwork
from utils.plotting import plot_loss, plot_gradient_heatmap, plot_convergence_comparison
print('imports ok')
```

```python
# Tiny training demo and save figures
rng = np.random.default_rng(1)
X = rng.normal(size=(3, 200))
true_w = rng.normal(size=(1,3))
y = (true_w @ X).reshape(1, -1)
net = NeuralNetwork(layer_sizes=[3, 16, 1], activation='relu', init_scheme='he', seed=0)
net.train(X, y, lr=0.05, epochs=120, verbose=False)
fig1 = plot_loss(net.loss_history)
fig2 = plot_gradient_heatmap(net.grad_norms_per_epoch)
from pathlib import Path
fig_dir = Path('figures')
fig_dir.mkdir(parents=True, exist_ok=True)
fig1.savefig(fig_dir / 'loss_curve.png', bbox_inches='tight')
fig2.savefig(fig_dir / 'grad_heatmap.png', bbox_inches='tight')
print('Saved loss_curve.png and grad_heatmap.png to', fig_dir)
```

```python
# Optional convergence experiment
try:
    from experiments.run_convergence import simple_compare
    acts = ['relu','gelu']
    inits = ['he','xavier']
    res = simple_compare(X, y, [3,16,1], acts, inits, lr=0.02, epochs=120)
    fig3 = plot_convergence_comparison(res)
    fig3.savefig(fig_dir / 'convergence_comparison.png', bbox_inches='tight')
    print('Saved convergence_comparison.png')
except Exception as e:
    print('Convergence experiment skipped or failed:', e)
```
